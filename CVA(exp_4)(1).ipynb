```python
import torch
import cv2
import matplotlib.pyplot as plt

# Load the YOLOv5 model (you can choose other models such as yolov5m, yolov5l for better accuracy)
model = torch.hub.load('ultralytics/yolov5', 'yolov5s')

# Path to the input video file (replace with your video file path)
video_path = ''  # Provide the correct path to your video file
cap = cv2.VideoCapture(video_path)  # Open video file

# Check if video opened successfully
if not cap.isOpened():
    print("Error: Could not open video.")
    exit()

# Loop through each frame of the video
while cap.isOpened():
    ret, frame = cap.read()  # Read the current frame
    if not ret:
        break  # Break if no frame is returned (end of video)

    # Perform object detection on the frame
    results = model(frame)  # Apply the YOLOv5 model to the frame

    # Render the results on the frame (bounding boxes, labels, confidence scores)
    results.render()  # This draws the boxes on the frame

    # Convert BGR (OpenCV format) to RGB (Matplotlib format)
    frame_with_detections = results.ims[0]  # The first image in the batch (as we only process one frame at a time)
    frame_with_detections_rgb = cv2.cvtColor(frame_with_detections, cv2.COLOR_BGR2RGB)

    # Display the frame with detections using Matplotlib
    plt.figure(figsize=(10, 6))
    plt.imshow(frame_with_detections_rgb)
    plt.axis('off')  # Hide axes for a cleaner display
    plt.show()

    # Optional: Slow down the loop for better viewing
    plt.pause(0.1)  # Adjust this value to control the speed of video playback

# Release the video capture object when done
cap.release()
```
